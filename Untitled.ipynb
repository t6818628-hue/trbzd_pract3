## 1.5. Все колонки датасета и их описание

## 1.6. Мода, среднее и медиана по Flight Distance

```python
col = 'Flight Distance'
print('Мода: ', ds[col].mode()[0])
print('Среднее: ', ds[col].mean())
print('Медиана: ', ds[col].median())
```

## 1.7. Уникальные значения в категориальных столбцах

```python
for c in ['Gender', 'Customer Type', 'Type of Travel', 'Class', 'Satisfaction']:
    print(c, ':', ds[c].unique())
```

## 1.8. Пропущенные значения (в %)

```python
ds.isnull().mean() * 100
```

Больше всего пропусков в `Arrival Delay` (для отмененных рейсов задержка прилета неизвестна).

## 1.9. Очистка NaN

```python
cat_cols = ['Gender', 'Customer Type', 'Type of Travel', 'Class', 'Satisfaction']
for c in cat_cols:
    ds[c] = ds[c].fillna('no data')
```

```python
num_cols = ds.select_dtypes(include=['int64', 'float64']).columns
ds[num_cols] = ds[num_cols].fillna(ds[num_cols].median())
```

Категориальные заменяем на `'no data'`, так как числа в текст писать нельзя. 
Числовые заменяем медианой, потому что в данных есть выбросы (задержки), а медиана к ним устойчива.

## 1.10-1.11. Проверка и удаление дубликатов

```python
print('Дубликатов:', ds.duplicated().sum())
print('До удаления:', ds.shape)
ds = ds.drop_duplicates()
print('После удаления:', ds.shape)
```

## 2.1. Коэффициент асимметрии и эксцесс (Flight Distance)

Выбираем `Flight Distance`, так как это непрерывный числовой признак с большим диапазоном значений.

```python
col = 'Flight Distance'
print('Skew: ', skew(ds[col]))
print('Kurt: ', kurtosis(ds[col]))
```

## 2.2. Сводная таблица и интерпретация

```python
stats = pd.DataFrame({
    "mean":   [ds[col].mean()],
    "median": [ds[col].median()],
    "std":    [ds[col].std()],
    "skew":   [skew(ds[col].dropna())],
    "kurt":   [kurtosis(ds[col].dropna())]
}).T

stats.columns = [col]
stats.index = ["Среднее", "Медиана", "Стандартное отклонение", "Асимметрия", "Эксцесс"]
stats
```

## 2.3. Коэффициент вариации (CV)

```python
cv = ds[col].std() / ds[col].mean() * 100
print('CV, %:', cv)
```

**Интерпретация:** Коэффициент вариации получился около 83% (значительно больше 33%). Это говорит о **высокой вариабельности**: дистанции полетов очень сильно разбросаны относительно среднего, данные неоднородны.

## 3.1. Фильтрация и сортировка

```python
filtered = ds.query("`Class` == 'Business' and `Departure Delay` > 60")
filtered.head()
```

**Вывод:** Отфильтрованы пассажиры бизнес-класса, чей рейс задержали более чем на час. Это недовольная аудитория, которая может негативно влиять на рейтинг авиакомпании.

```python
sort = ds.sort_values(by='Flight Distance', ascending=False)
sort.head()
```

**Вывод:** Отсортировали данные, чтобы увидеть самые длинные перелеты (топ-5). Это позволяет выявить экстремальные значения.

## 3.2. Срезы данных

```python
print("--- loc ---")
display(ds.loc[0:5, ['Age', 'Flight Distance', 'Satisfaction']])

print("--- loc с условием ---")
display(ds.loc[ds['Satisfaction'] == 'Satisfied', ['Age', 'Class', 'Flight Distance']].head())

print("--- iloc ---")
display(ds.iloc[0:5, 0:3])
```

## 3.3. Группировка данных

```python
ds.groupby('Class')[['Seat Comfort', 'Food and Drink', 'In-flight Entertainment', 'Flight Distance']].mean()
```

**Вывод:** Группировка по классу обслуживания показывает, что в бизнес-классе оценки комфорта, еды и развлечений значительно выше, чем в экономе. Также средняя дистанция полета бизнес-класса больше, что логично.

## 4. Топ-5 записей

```python
top5 = ds.sort_values(by='Flight Distance', ascending=False).head(5)
top5_df = pd.DataFrame(top5)
top5_df
```

Рейтинг показывает, что даже на таких длинных рейсах есть пассажиры, которые остались довольны, что говорит о высоком уровне сервиса на дальних направлениях.
